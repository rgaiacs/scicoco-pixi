```python
import requests
from PIL import Image
from io import BytesIO
```

```python
with requests.get(
    'https://github.com/python-pillow/Pillow/' +
    'blob/main/Tests/images/hopper.jpg?raw=true'
    ) as r:
    if r.status_code == 200:
        hopper = Image.open(BytesIO(r.content))

hopper
```
